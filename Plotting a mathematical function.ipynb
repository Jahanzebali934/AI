```python
import matplotlib.pyplot as plt
import numpy as np

# Create 100 values between -3 and 3
x = np.linspace(-3, 3, 100)

# Apply the given function
y = np.maximum(x, 0)

# Draw the graph
plt.plot(x, y)

# Add labels and title
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Plot of the ReLU Function")

# Display grid
plt.grid(True)

# Show the graph
plt.show()
```
